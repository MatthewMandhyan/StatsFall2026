```python
import numpy as np
 
def G(s, lam):
    return np.exp(lam * (s - 1))
 
def dG(s, lam):
    return lam * np.exp(lam * (s - 1))
 
def recursion(lam, tol=1e-8, max_iter=100000):
    a = 0.0
    for n in range(1, max_iter + 1):
        new = G(a, lam)
        if abs(new - a) < tol:
            return new, n
        a = new
    return a, max_iter
 
def newton(lam, tol=1e-8, max_iter=1000):
    s = 0.0
    for k in range(1, max_iter + 1):
        new = s - (G(s, lam) - s) / (dG(s, lam) - 1)
        if abs(new - s) < tol:
            return new, k
        s = new
    return s, max_iter
 
print(f"{'lambda'} {'recursion alpha'} {'iters'} {'newton alpha'} {'iters'}")
for lam in [0.5, 1, 1.5, 2]:
    a_r, n_r = recursion(lam)
    a_n, n_n = newton(lam)
    print(f"{lam} {a_r} {n_r} {a_n} {n_n}")
```
