```python
import numpy as np
import matplotlib.pyplot as plt
 
p = np.linspace(0.01, 0.99, 500)
q = 1 - p
 
def G(s):
    return (q + p * s) ** 2
 
alpha = np.minimum(1, (q / p) ** 2)
a1 = G(0)
a2 = G(a1)
a3 = G(a2)
 
plt.plot(p, alpha, label="alpha")
plt.plot(p, a1, label="alpha(1)")
plt.plot(p, a2, label="alpha(2)")
plt.plot(p, a3, label="alpha(3)")
plt.xlabel("p")
plt.ylabel("probability")
plt.title("Extinction probabilities, Binomial(2, p) offspring")
plt.legend()
plt.grid(True)
plt.show()
```
