```python
import numpy as np
import matplotlib.pyplot as plt
 
M = 5
delta = 0.5
p = np.linspace(0.01, 0.99, 500)
r = (1 - p) / p
 
denom = 1 + (1 - delta) * sum(r**i for i in range(1, M))
 
for j in range(1, M):
    num = 1 + (1 - delta) * sum(r**i for i in range(1, j))
    plt.plot(p, num / denom, label=f"alpha_{j}")
 
plt.xlabel("p")
plt.ylabel("alpha_j")
plt.title("Absorption probabilities, M = 5, delta = 0.5")
plt.legend()
plt.grid(True)
plt.show()
```
