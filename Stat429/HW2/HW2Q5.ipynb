```python
import numpy as np
import numpy.linalg as la
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm
import librosa
from scipy.special import softmax
from scipy import stats
```

```python
halloween = pd.read_csv("time_series_US_20031231-1800_20261002-1900.csv")
halloween.head()
```

```python
halloween2 = halloween.iloc[:-21]
n = len(halloween2)
t = np.arange(n)
plt.figure(figsize=(24,12))
plt.plot(t,halloween2['halloween'])
plt.xlabel('Months since Jan 2004')
plt.ylabel('Search interest')
plt.show()
```

The data has a rythm that seems pretty evident. This logically makes sense as the interest in halloween tends to come around the time of october each year. Noting that 2004-2024 is 20 years and we see 20 peaks. 

Part C

```python
fft_vals = np.fft.fft(halloween2['halloween'])
periodogram = np.abs(fft_vals ** 2)/n
plt.figure(figsize=(24,12))
plt.plot(t, periodogram)
plt.xlabel('Months since Jan 2004')
plt.ylabel('Periodogram')
plt.show()
```

These results align with part a as we would expect this evidently periodic. And in accordance to the results in part a we see that when the periodogram takes a value >0 happens in an evenly spaced manner. Thus like in a we suspect that this is happening with respect to some factor k.
